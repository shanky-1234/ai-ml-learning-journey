# Data Cleaning

```python
df_clean = df.copy()
```

```python
df_clean.drop(columns=['playlist_name', 'track_artist','track_album_id','playlist_id','track_id','track_album_name','track_name', 'track_href','track_album_release_date','analysis_url','id','uri','type'], axis=1,inplace=True)
```

```python
df_clean
```

```python
df_clean.drop('mode',axis=1,inplace=True)
```

# Feature Extraction


```python
# Tempo Category 
def temp_category(tempo):
    if tempo < 90:
        return 'slow'
    elif tempo >= 90 and tempo <= 120:
        return 'medium'
    else:
        return 'fast'

df_clean['tempo_category'] = df_clean['tempo'].apply(temp_category)
```

```python
# Tempo Category 
def popularity(track_popularity):
    if track_popularity < 40:
        return 'less'
    elif track_popularity >= 40 and track_popularity <= 70:
        return 'average'
    else:
        return 'popular'

df_clean['popularity'] = df_clean['track_popularity'].apply(popularity)
```

```python
df_clean
```

# Splitting the model

```python

X = df_clean.drop(['track_popularity','popularity'],axis=1)
y=df_clean['popularity']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.preprocessing import OneHotEncoder
oneHot = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
```

```python
X_train_OneHot = oneHot.fit_transform(X_train[['playlist_genre']])
X_test_OneHot = oneHot.transform(X_test[['playlist_genre']])

X_train_OneHot_df = pd.DataFrame(X_train_OneHot,columns = oneHot.get_feature_names_out(['playlist_genre']),index=X_train.index)
X_test_OneHot_df = pd.DataFrame(X_test_OneHot,columns = oneHot.get_feature_names_out(['playlist_genre']),index=X_test.index)
```

```python
X_train = X_train.drop('playlist_genre',axis=1)
X_train = pd.concat([X_train,X_train_OneHot_df],axis=1)

X_test = X_test.drop('playlist_genre',axis=1)
X_test = pd.concat([X_test,X_test_OneHot_df],axis=1)
```

```python
df_clean['playlist_subgenre'].value_counts()
```

```python
X_train_OneHot = oneHot.fit_transform(X_train[['playlist_subgenre']])
X_test_OneHot = oneHot.transform(X_test[['playlist_subgenre']])

X_train_OneHot_df = pd.DataFrame(X_train_OneHot,columns = oneHot.get_feature_names_out(['playlist_subgenre']),index=X_train.index)
X_test_OneHot_df = pd.DataFrame(X_test_OneHot,columns = oneHot.get_feature_names_out(['playlist_subgenre']),index=X_test.index)

```

```python
X_train = X_train.drop('playlist_subgenre',axis=1)
X_train = pd.concat([X_train,X_train_OneHot_df],axis=1)

X_test = X_test.drop('playlist_subgenre',axis=1)
X_test = pd.concat([X_test,X_test_OneHot_df],axis=1)
```

```python
X_train
```

Since Time Signature and key are also categorical let's encode it 

```python
X_train_OneHot = oneHot.fit_transform(X_train[['time_signature','key']])
X_test_OneHot = oneHot.transform(X_test[['time_signature','key']])

X_train_OneHot_df = pd.DataFrame(X_train_OneHot,columns = oneHot.get_feature_names_out(['time_signature','key']),index=X_train.index)
X_test_OneHot_df = pd.DataFrame(X_test_OneHot,columns = oneHot.get_feature_names_out(['time_signature','key']),index=X_test.index)

X_train = X_train.drop(['time_signature','key'],axis=1)
X_train = pd.concat([X_train,X_train_OneHot_df],axis=1)

X_test = X_test.drop(['time_signature','key'],axis=1)
X_test = pd.concat([X_test,X_test_OneHot_df],axis=1)
```

```python
# Label Emcoding
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
X_train['tempo_category'] = encoder.fit_transform(X_train['tempo_category'])
X_test['tempo_category'] = encoder.transform(X_test['tempo_category'])
```

# Model Training
We will test through various methods to find the best model

**Supervised Learning**

First Let's try with logistic regression but with softmax

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(max_iter=1000)
```

```python
logistic_model.fit(X_train_scaled,y_train)
```

```python
pred = logistic_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,pred)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_test,pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, pred))
```

```python
# Hyperparameter Tuning
from sklearn.model_selection import GridSearchCV

params = {
    'C': [0.01, 1, 10, 100],
    'solver': ['lbfgs', 'liblinear'],
    'penalty': ['l2']
}
gridSearch = GridSearchCV((logistic_model),params,cv=5,return_train_score=False)
```

```python
gridSearch.fit(X_train_scaled,y_train)
```

```python
gridSearch.score(X_test_scaled,y_test)
```
